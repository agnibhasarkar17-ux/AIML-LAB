```python
#Problem 2: The Water Jug Problem (Solved using BFS)
from collections import deque

capacity_A = 4
capacity_B = 3


def water_jug_bfs():
    # Initial state: both jugs are empty
    start = (0, 0)

    queue = deque([(start, [])])
    visited = {start}

    while queue:
        (a, b), path = queue.popleft()

        # Check whether we have exactly 2 litres
        if a == 2 or b == 2:
            return path + [(a, b)]

        next_states = []

        # 1. Fill Jug A
        next_states.append(((capacity_A, b), "Fill Jug A"))

        # 2. Fill Jug B
        next_states.append(((a, capacity_B), "Fill Jug B"))

        # 3. Empty Jug A
        next_states.append(((0, b), "Empty Jug A"))

        # 4. Empty Jug B
        next_states.append(((a, 0), "Empty Jug B"))

        # 5. Pour A -> B
        amount = min(a, capacity_B - b)
        next_states.append(
            ((a - amount, b + amount), "Pour A -> B")
        )

        # 6. Pour B -> A
        amount = min(b, capacity_A - a)
        next_states.append(
            ((a + amount, b - amount), "Pour B -> A")
        )

        for state, action in next_states:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [action]))

    return None


solution = water_jug_bfs()

print("Minimum sequence of steps:")

for i, step in enumerate(solution):
    if i == 0:
        print(f"Step {i}: Start with {step}")
    else:
        print(f"Step {i}: {step}")
```
