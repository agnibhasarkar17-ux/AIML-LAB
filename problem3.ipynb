```python
# Road network represented using an adjacency list

graph = {
    'Kolkata': ['Delhi', 'Mumbai'],
    'Delhi': ['Kolkata', 'Jaipur'],
    'Mumbai': ['Kolkata', 'Pune'],
    'Jaipur': ['Delhi'],
    'Pune': ['Mumbai'],
    'Chennai': ['Bangalore'],
    'Bangalore': ['Chennai']
}


# DFS to check whether a route exists
def route_exists(graph, start, destination):

    visited = set()

    def dfs(city):
        if city == destination:
            return True

        visited.add(city)

        for neighbour in graph.get(city, []):
            if neighbour not in visited:
                if dfs(neighbour):
                    return True

        return False

    return dfs(start)


# DFS to find one connected component
def find_component(graph, start, visited):

    component = []

    def dfs(city):
        visited.add(city)
        component.append(city)

        for neighbour in graph.get(city, []):
            if neighbour not in visited:
                dfs(neighbour)

    dfs(start)

    return component


# Find all connected components
def connected_components(graph):

    visited = set()
    components = []

    for city in graph:

        if city not in visited:
            component = find_component(
                graph, city, visited
            )

            components.append(component)

    return components


# Check routes
print("Route exists:")
print("Kolkata -> Pune:",
      route_exists(graph, 'Kolkata', 'Pune'))

print("Kolkata -> Chennai:",
      route_exists(graph, 'Kolkata', 'Chennai'))


# Find connected components
print("\nConnected Components:")

components = connected_components(graph)

for i, component in enumerate(components, 1):
    print(f"Component {i}: {component}")
```
